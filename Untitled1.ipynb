```python

# ============================================================
# ПОЛНЫЙ АНАЛИЗ TELEGRAM-КАНАЛА МОСКОВСКОГО УНИВЕРСИТЕТА МВД
# ============================================================

# 1. ИМПОРТ БИБЛИОТЕК
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

import csv
import re
import ast
import os
import warnings

from collections import Counter
from scipy import stats
from IPython.display import display

warnings.filterwarnings('ignore')

# ============================================================
# 2. ЗАГРУЗКА ДАННЫХ
# ============================================================

FILE_PATH = '/content/mvd.csv'

if not os.path.exists(FILE_PATH):
    raise FileNotFoundError(
        f'Файл не найден: {FILE_PATH}. '
        'Проверь, загружен ли mvd.csv в Colab.'
    )

# Определяем разделитель CSV
with open(FILE_PATH, 'r', encoding='utf-8-sig') as f:
    sample = f.read(20000)

try:
    delimiter = csv.Sniffer().sniff(
        sample, delimiters=';,\t|'
    ).delimiter
except:
    delimiter = ';'

df = pd.read_csv(
    FILE_PATH,
    sep=delimiter,
    encoding='utf-8-sig',
    engine='python',
    on_bad_lines='warn'
)

# Очищаем названия столбцов
df.columns = (
    df.columns.astype(str)
    .str.strip()
    .str.lower()
)

print('=' * 60)
print('ЗАГРУЗКА ДАННЫХ')
print('=' * 60)

print('Размер таблицы:', df.shape)
print('\nСтолбцы:')
print(df.columns.tolist())

print('\nПервые пять строк:')
display(df.head())

# ============================================================
# 3. ПРОВЕРКА И ПОДГОТОВКА СТОЛБЦОВ
# ============================================================

required_columns = [
    'date',
    'message_id',
    'message',
    'views',
    'comments',
    'reactions',
    'channel_members_count',
    'reactions_total',
    'forwarded',
    'forwards',
    'topic_title'
]

for col in required_columns:
    if col not in df.columns:
        df[col] = np.nan

# Преобразование числовых показателей
numeric_columns = [
    'views',
    'comments',
    'reactions_total',
    'forwards',
    'channel_members_count'
]

for col in numeric_columns:
    df[col] = pd.to_numeric(
        df[col], errors='coerce'
    )

# Если reactions_total отсутствует,
# попробуем рассчитать его из reactions
def count_reactions(value):
    if pd.isna(value):
        return 0

    if isinstance(value, dict):
        return sum(
            pd.to_numeric(v, errors='coerce') or 0
            for v in value.values()
        )

    if isinstance(value, list):
        total = 0
        for item in value:
            if isinstance(item, dict):
                count = item.get('count', 0)
                total += pd.to_numeric(
                    count, errors='coerce'
                ) or 0
        return total

    if isinstance(value, str):
        try:
            parsed = ast.literal_eval(value)
            return count_reactions(parsed)
        except:
            return 0

    return 0


if df['reactions_total'].isna().all():
    df['reactions_total'] = df['reactions'].apply(
        count_reactions
    )
else:
    missing_reactions = df['reactions_total'].isna()
    df.loc[missing_reactions, 'reactions_total'] = (
        df.loc[missing_reactions, 'reactions']
        .apply(count_reactions)
    )

# Пропущенные значения в количественных показателях
for col in numeric_columns:
    df[col] = df[col].fillna(0)

# Приводим текстовые поля к строкам
df['message'] = df['message'].fillna('').astype(str)
df['topic_title'] = df['topic_title'].fillna('Не указано').astype(str)

# Обрабатываем даты.
# Время без часового пояса интерпретируется как UTC.
df['date'] = pd.to_datetime(
    df['date'],
    errors='coerce',
    utc=True
)

# Удаляем строки без даты
df = df.dropna(subset=['date']).copy()

# Удаляем дубли по ID публикации, если он есть
if df['message_id'].notna().any():
    df = df.drop_duplicates(
        subset=['message_id'],
        keep='first'
    )

# ============================================================
# 4. СОЗДАНИЕ ДОПОЛНИТЕЛЬНЫХ ПОКАЗАТЕЛЕЙ
# ============================================================

# Временные характеристики
df['year'] = df['date'].dt.year
df['month'] = df['date'].dt.month
df['month_name'] = df['date'].dt.strftime('%Y-%m')
df['hour'] = df['date'].dt.hour
df['weekday'] = df['date'].dt.dayofweek

weekday_names = {
    0: 'Понедельник',
    1: 'Вторник',
    2: 'Среда',
    3: 'Четверг',
    4: 'Пятница',
    5: 'Суббота',
    6: 'Воскресенье'
}

df['weekday_name'] = df['weekday'].map(weekday_names)

# Длина текста
df['text_length'] = df['message'].str.len()

# Количество слов
df['word_count'] = df['message'].apply(
    lambda x: len(re.findall(r'\b\w+\b', x, flags=re.UNICODE))
)

# Наличие текста
df['has_text'] = df['message'].str.strip().ne('')

# Определение медиа
df['is_media'] = (
    df['message'].str.contains(
        'MediaMessage',
        case=False,
        na=False
    )
)

df['publication_format'] = np.where(
    df['is_media'],
    'Медиа',
    np.where(df['has_text'], 'Текст', 'Без текста')
)

# ER — реакции и комментарии относительно просмотров
df['engagement'] = (
    df['reactions_total'] + df['comments']
)

df['ER_percent'] = np.where(
    df['views'] > 0,
    df['engagement'] / df['views'] * 100,
    np.nan
)

# Доля просмотров относительно количества подписчиков.
# Это не уникальный охват, а отношение просмотров к подписчикам.
df['reach_percent'] = np.where(
    df['channel_members_count'] > 0,
    df['views'] / df['channel_members_count'] * 100,
    np.nan
)

# Доля пересылок относительно просмотров
df['forward_rate'] = np.where(
    df['views'] > 0,
    df['forwards'] / df['views'] * 100,
    np.nan
)

# Проверяем признак пересылки
def to_bool(value):
    if pd.isna(value):
        return False
    if isinstance(value, bool):
        return value
    return str(value).strip().lower() in [
        'true', '1', 'yes', 'да'
    ]

df['forwarded_bool'] = df['forwarded'].apply(to_bool)

print('\nДанные подготовлены.')
print('Количество публикаций:', len(df))

# ============================================================
# 5. ОБЩАЯ СТАТИСТИКА КАНАЛА
# ============================================================

print('\n' + '=' * 60)
print('1. ОБЩАЯ СТАТИСТИКА')
print('=' * 60)

general_stats = pd.DataFrame({
    'Показатель': [
        'Количество публикаций',
        'Среднее количество просмотров',
        'Медианное количество просмотров',
        'Максимальное количество просмотров',
        'Среднее количество реакций',
        'Среднее количество комментариев',
        'Среднее количество пересылок',
        'Средний ER, %',
        'Средний показатель просмотров к подписчикам, %',
        'Средняя длина публикации, символов',
        'Дата первой публикации',
        'Дата последней публикации'
    ],
    'Значение': [
        len(df),
        round(df['views'].mean(), 2),
        round(df['views'].median(), 2),
        df['views'].max(),
        round(df['reactions_total'].mean(), 2),
        round(df['comments'].mean(), 2),
        round(df['forwards'].mean(), 2),
        round(df['ER_percent'].mean(), 2),
        round(df['reach_percent'].mean(), 2),
        round(df['text_length'].mean(), 2),
        df['date'].min(),
        df['date'].max()
    ]
})

display(general_stats)

# ============================================================
# 6. ТОП ПУБЛИКАЦИЙ
# ============================================================

print('\n' + '=' * 60)
print('2. ТОП ПУБЛИКАЦИЙ')
print('=' * 60)

top_columns = [
    'date',
    'message',
    'views',
    'reactions_total',
    'comments',
    'forwards',
    'ER_percent'
]

top_views = df.nlargest(10, 'views')[top_columns]
top_reactions = df.nlargest(10, 'reactions_total')[top_columns]
top_forwards = df.nlargest(10, 'forwards')[top_columns]
top_er = df.nlargest(10, 'ER_percent')[top_columns]

print('\nТОП-10 по просмотрам:')
display(top_views)

print('\nТОП-10 по реакциям:')
display(top_reactions)

print('\nТОП-10 по пересылкам:')
display(top_forwards)

print('\nТОП-10 по ER:')
display(top_er)

# ============================================================
# 7. АНАЛИЗ ПУБЛИКАЦИЙ ПО МЕСЯЦАМ
# ============================================================

print('\n' + '=' * 60)
print('3. ДИНАМИКА ПО МЕСЯЦАМ')
print('=' * 60)

monthly_stats = (
    df.groupby('month_name')
    .agg(
        publications=('message_id', 'count'),
        average_views=('views', 'mean'),
        median_views=('views', 'median'),
        total_views=('views', 'sum'),
        average_reactions=('reactions_total', 'mean'),
        average_comments=('comments', 'mean'),
        average_forwards=('forwards', 'mean'),
        average_ER=('ER_percent', 'mean')
    )
    .round(2)
    .reset_index()
)

display(monthly_stats)

plt.figure(figsize=(14, 5))
plt.plot(
    monthly_stats['month_name'],
    monthly_stats['average_views'],
    marker='o'
)
plt.title('Среднее количество просмотров по месяцам')
plt.xlabel('Месяц')
plt.ylabel('Среднее количество просмотров')
plt.xticks(rotation=45)
plt.grid(alpha=0.3)
plt.tight_layout()
plt.show()

plt.figure(figsize=(14, 5))
plt.bar(
    monthly_stats['month_name'],
    monthly_stats['publications']
)
plt.title('Количество публикаций по месяцам')
plt.xlabel('Месяц')
plt.ylabel('Количество публикаций')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()

# ============================================================
# 8. АНАЛИЗ ВРЕМЕНИ ПУБЛИКАЦИЙ
# ============================================================

print('\n' + '=' * 60)
print('4. ВРЕМЯ ПУБЛИКАЦИЙ')
print('=' * 60)

hour_stats = (
    df.groupby('hour')
    .agg(
        publications=('message_id', 'count'),
        average_views=('views', 'mean'),
        average_ER=('ER_percent', 'mean')
    )
    .round(2)
    .reset_index()
)

weekday_stats = (
    df.groupby(['weekday', 'weekday_name'])
    .agg(
        publications=('message_id', 'count'),
        average_views=('views', 'mean'),
        average_ER=('ER_percent', 'mean')
    )
    .round(2)
    .reset_index()
    .sort_values('weekday')
)

print('\nСтатистика по часам:')
display(hour_stats)

print('\nСтатистика по дням недели:')
display(weekday_stats)

plt.figure(figsize=(12, 5))
sns.barplot(
    data=hour_stats,
    x='hour',
    y='average_views'
)
plt.title('Средние просмотры в зависимости от часа публикации')
plt.xlabel('Час (UTC)')
plt.ylabel('Средние просмотры')
plt.tight_layout()
plt.show()

plt.figure(figsize=(12, 5))
sns.barplot(
    data=weekday_stats,
    x='weekday_name',
    y='average_views'
)
plt.title('Средние просмотры по дням недели')
plt.xlabel('День недели')
plt.ylabel('Средние просмотры')
plt.xticks(rotation=30)
plt.tight_layout()
plt.show()

# Тепловая карта: день недели × час
heatmap_data = df.pivot_table(
    index='weekday_name',
    columns='hour',
    values='views',
    aggfunc='mean'
)

weekday_order = [
    'Понедельник', 'Вторник', 'Среда',
    'Четверг', 'Пятница', 'Суббота',
    'Воскресенье'
]

heatmap_data = heatmap_data.reindex(weekday_order)

plt.figure(figsize=(16, 6))
sns.heatmap(
    heatmap_data,
    cmap='YlOrRd',
    annot=False
)
plt.title('Средние просмотры: день недели и час публикации')
plt.xlabel('Час (UTC)')
plt.ylabel('День недели')
plt.tight_layout()
plt.show()

# ============================================================
# 9. АНАЛИЗ ФОРМАТА ПУБЛИКАЦИЙ
# ============================================================

print('\n' + '=' * 60)
print('5. ФОРМАТЫ ПУБЛИКАЦИЙ')
print('=' * 60)

format_stats = (
    df.groupby('publication_format')
    .agg(
        publications=('message_id', 'count'),
        average_views=('views', 'mean'),
        median_views=('views', 'median'),
        average_reactions=('reactions_total', 'mean'),
        average_comments=('comments', 'mean'),
        average_forwards=('forwards', 'mean'),
        average_ER=('ER_percent', 'mean')
    )
    .round(2)
    .reset_index()
)

display(format_stats)

plt.figure(figsize=(8, 5))
sns.barplot(
    data=format_stats,
    x='publication_format',
    y='average_views'
)
plt.title('Средние просмотры по формату публикаций')
plt.xlabel('Формат')
plt.ylabel('Средние просмотры')
plt.tight_layout()
plt.show()

# ============================================================
# 10. АНАЛИЗ ДЛИНЫ ТЕКСТА
# ============================================================

print('\n' + '=' * 60)
print('6. ДЛИНА ТЕКСТА')
print('=' * 60)

df['text_length_group'] = pd.cut(
    df['text_length'],
    bins=[-1, 100, 300, 600, 1000, np.inf],
    labels=[
        'До 100',
        '101–300',
        '301–600',
        '601–1000',
        'Более 1000'
    ]
)

length_stats = (
    df.groupby('text_length_group', observed=False)
    .agg(
        publications=('message_id', 'count'),
        average_views=('views', 'mean'),
        median_views=('views', 'median'),
        average_ER=('ER_percent', 'mean')
    )
    .round(2)
    .reset_index()
)

display(length_stats)

plt.figure(figsize=(10, 5))
sns.barplot(
    data=length_stats,
    x='text_length_group',
    y='average_views'
)
plt.title('Средние просмотры в зависимости от длины текста')
plt.xlabel('Длина текста, символов')
plt.ylabel('Средние просмотры')
plt.tight_layout()
plt.show()

# ============================================================
# 11. АНАЛИЗ СЛОВ И ХЕШТЕГОВ
# ============================================================

print('\n' + '=' * 60)
print('7. ЧАСТОТНЫЙ АНАЛИЗ ТЕКСТОВ')
print('=' * 60)

stop_words = {
    'и', 'в', 'во', 'на', 'с', 'со', 'по', 'к', 'у',
    'из', 'за', 'для', 'от', 'до', 'о', 'об', 'при',
    'а', 'но', 'или', 'что', 'как', 'это', 'так',
    'его', 'ее', 'их', 'мы', 'вы', 'они', 'он', 'она',
    'не', 'да', 'нет', 'бы', 'же', 'ли', 'то', 'все',
    'всех', 'был', 'была', 'были', 'будет', 'быть',
    'который', 'которая', 'которые', 'этот', 'эта',
    'эти', 'также', 'уже', 'ещё', 'еще', 'после',
    'перед', 'через', 'под', 'над', 'между'
}

all_text = ' '.join(df['message'].tolist()).lower()

words = re.findall(
    r'[а-яёa-z]{3,}',
    all_text,
    flags=re.IGNORECASE
)

filtered_words = [
    word for word in words
    if word not in stop_words
]

word_frequency = Counter(filtered_words)

top_words = pd.DataFrame(
    word_frequency.most_common(30),
    columns=['Слово', 'Частота']
)

print('\nТОП-30 часто используемых слов:')
display(top_words)

# Хештеги
hashtags = re.findall(
    r'#[а-яёa-z0-9_]+',
    all_text,
    flags=re.IGNORECASE
)

hashtag_frequency = Counter(hashtags)

top_hashtags = pd.DataFrame(
    hashtag_frequency.most_common(30),
    columns=['Хештег', 'Частота']
)

print('\nТОП-30 хештегов:')
display(top_hashtags)

# ============================================================
# 12. АНАЛИЗ ТЕМ ПУБЛИКАЦИЙ
# ============================================================

print('\n' + '=' * 60)
print('8. АНАЛИЗ ТЕМ')
print('=' * 60)

topic_stats = (
    df.groupby('topic_title')
    .agg(
        publications=('message_id', 'count'),
        average_views=('views', 'mean'),
        median_views=('views', 'median'),
        average_reactions=('reactions_total', 'mean'),
        average_ER=('ER_percent', 'mean'),
        average_forwards=('forwards', 'mean')
    )
    .round(2)
    .reset_index()
    .sort_values('publications', ascending=False)
)

display(topic_stats.head(30))

# ============================================================
# 13. АНАЛИЗ ПЕРЕСЫЛОК
# ============================================================

print('\n' + '=' * 60)
print('9. АНАЛИЗ ПЕРЕСЫЛОК')
print('=' * 60)

forward_stats = (
    df.groupby('forwarded_bool')
    .agg(
        publications=('message_id', 'count'),
        average_views=('views', 'mean'),
        average_reactions=('reactions_total', 'mean'),
        average_forwards=('forwards', 'mean'),
        average_ER=('ER_percent', 'mean')
    )
    .round(2)
    .reset_index()
)

display(forward_stats)

# ============================================================
# 14. КОРРЕЛЯЦИОННЫЙ АНАЛИЗ
# ============================================================

print('\n' + '=' * 60)
print('10. КОРРЕЛЯЦИОННЫЙ АНАЛИЗ')
print('=' * 60)

correlation_columns = [
    'views',
    'comments',
    'reactions_total',
    'forwards',
    'text_length',
    'word_count',
    'ER_percent',
    'reach_percent'
]

correlation_data = df[
    correlation_columns
].replace([np.inf, -np.inf], np.nan)

correlation_matrix = correlation_data.corr(
    method='spearman'
)

display(correlation_matrix.round(3))

plt.figure(figsize=(10, 8))
sns.heatmap(
    correlation_matrix,
    annot=True,
    cmap='coolwarm',
    center=0,
    fmt='.2f'
)
plt.title('Корреляционная матрица Спирмена')
plt.tight_layout()
plt.show()

# ============================================================
# 15. ПРОВЕРКА СТАТИСТИЧЕСКИХ ГИПОТЕЗ
# ============================================================

print('\n' + '=' * 60)
print('11. СТАТИСТИЧЕСКИЕ ГИПОТЕЗЫ')
print('=' * 60)

# Гипотеза 1:
# Отличаются ли просмотры текстовых и медийных публикаций?

text_views = df.loc[
    df['publication_format'] == 'Текст',
    'views'
].dropna()

media_views = df.loc[
    df['publication_format'] == 'Медиа',
    'views'
].dropna()

if len(text_views) >= 2 and len(media_views) >= 2:
    u_stat, p_value = stats.mannwhitneyu(
        text_views,
        media_views,
        alternative='two-sided'
    )

    print('\nГипотеза 1: текст против медиа')
    print('Медиана просмотров текста:', text_views.median())
    print('Медиана просмотров медиа:', media_views.median())
    print('U-статистика:', u_stat)
    print('p-value:', p_value)

    if p_value < 0.05:
        print('Результат: различие статистически значимо.')
    else:
        print('Результат: статистически значимого различия не обнаружено.')
else:
    print('\nНедостаточно данных для сравнения форматов.')

# Гипотеза 2:
# Различаются ли просмотры в зависимости от дня недели?

weekday_groups = [
    group['views'].dropna().values
    for _, group in df.groupby('weekday')
    if len(group) >= 2
]

if len(weekday_groups) >= 2:
    h_stat, p_value = stats.kruskal(*weekday_groups)

    print('\nГипотеза 2: различия по дням недели')
    print('H-статистика:', h_stat)
    print('p-value:', p_value)

    if p_value < 0.05:
        print('Результат: есть статистически значимые различия.')
    else:
        print('Результат: значимых различий не обнаружено.')
else:
    print('\nНедостаточно данных для сравнения дней недели.')

# Гипотеза 3:
# Связаны ли просмотры с количеством реакций?

valid_corr = df[
    ['views', 'reactions_total']
].dropna()

if len(valid_corr) >= 3:
    corr, p_value = stats.spearmanr(
        valid_corr['views'],
        valid_corr['reactions_total']
    )

    print('\nГипотеза 3: просмотры и реакции')
    print('Коэффициент Спирмена:', round(corr, 3))
    print('p-value:', p_value)

    if p_value < 0.05:
        print('Результат: связь статистически значима.')
    else:
        print('Результат: статистически значимая связь не обнаружена.')

# ============================================================
# 16. РЕЙТИНГ ПУБЛИКАЦИЙ
# ============================================================

print('\n' + '=' * 60)
print('12. РЕЙТИНГ ПУБЛИКАЦИЙ')
print('=' * 60)

# Нормализация показателей от 0 до 1
def normalize(series):
    minimum = series.min()
    maximum = series.max()

    if maximum == minimum:
        return pd.Series(0, index=series.index)

    return (series - minimum) / (maximum - minimum)


df['views_norm'] = normalize(df['views'])
df['reactions_norm'] = normalize(df['reactions_total'])
df['forwards_norm'] = normalize(df['forwards'])

# Условный интегральный индекс.
# Вес просмотров — 50%, реакций — 30%, пересылок — 20%.
df['publication_score'] = (
    df['views_norm'] * 0.5
    + df['reactions_norm'] * 0.3
    + df['forwards_norm'] * 0.2
)

rating = df.nlargest(
    20, 'publication_score'
)[
    [
        'date',
        'message',
        'views',
        'reactions_total',
        'forwards',
        'ER_percent',
        'publication_score'
    ]
]

display(rating)

# ============================================================
# 17. СОХРАНЕНИЕ РЕЗУЛЬТАТОВ В EXCEL
# ============================================================

print('\n' + '=' * 60)
print('13. ЭКСПОРТ РЕЗУЛЬТАТОВ')
print('=' * 60)

OUTPUT_FILE = '/content/mvd_analysis_results.xlsx'

# Убираем часовой пояс из дат для совместимости с Excel
df_export = df.copy()

for col in df_export.columns:
    if isinstance(df_export[col].dtype, pd.DatetimeTZDtype):
        df_export[col] = df_export[col].dt.tz_localize(None)

with pd.ExcelWriter(
    OUTPUT_FILE,
    engine='openpyxl'
) as writer:

    general_stats.to_excel(
        writer, sheet_name='Общая статистика', index=False
    )

    df_export.to_excel(
        writer, sheet_name='Все публикации', index=False
    )

    top_views.to_excel(
        writer, sheet_name='Топ просмотров', index=False
    )

    top_reactions.to_excel(
        writer, sheet_name='Топ реакций', index=False
    )

    top_forwards.to_excel(
        writer, sheet_name='Топ пересылок', index=False
    )

    top_er.to_excel(
        writer, sheet_name='Топ ER', index=False
    )

    monthly_stats.to_excel(
        writer, sheet_name='Динамика по месяцам', index=False
    )

    hour_stats.to_excel(
        writer, sheet_name='По часам', index=False
    )

    weekday_stats.to_excel(
        writer, sheet_name='По дням недели', index=False
    )

    format_stats.to_excel(
        writer, sheet_name='Форматы', index=False
    )

    length_stats.to_excel(
        writer, sheet_name='Длина текста', index=False
    )

    top_words.to_excel(
        writer, sheet_name='Частые слова', index=False
    )

    top_hashtags.to_excel(
        writer, sheet_name='Хештеги', index=False
    )

    topic_stats.to_excel(
        writer, sheet_name='Темы', index=False
    )

    forward_stats.to_excel(
        writer, sheet_name='Пересылки', index=False
    )

    correlation_matrix.to_excel(
        writer, sheet_name='Корреляции'
    )

    rating.to_excel(
        writer, sheet_name='Рейтинг публикаций', index=False
    )

print('Анализ завершён!')
print('Excel-файл сохранён:', OUTPUT_FILE)

# Автоматическое скачивание файла в Google Colab
from google.colab import files
files.download(OUTPUT_FILE)
```

```python

# ============================================================
# ДОПОЛНИТЕЛЬНЫЙ УГЛУБЛЕННЫЙ АНАЛИЗ TELEGRAM-КАНАЛА
# ============================================================

import pandas as pd
import numpy as np
import re
import os
import warnings
import matplotlib.pyplot as plt
import seaborn as sns

from collections import Counter
from scipy import stats
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.cluster import KMeans
from sklearn.ensemble import RandomForestRegressor
from sklearn.preprocessing import StandardScaler
from sklearn.impute import SimpleImputer
from sklearn.pipeline import Pipeline
from sklearn.metrics import r2_score, mean_absolute_error

warnings.filterwarnings('ignore')

# ============================================================
# 0. ПРОВЕРКА ДАННЫХ
# ============================================================

if 'df' not in globals():
    FILE_PATH = '/content/mvd.csv'
    if not os.path.exists(FILE_PATH):
        raise FileNotFoundError('Не найден файл /content/mvd.csv')

    import csv
    with open(FILE_PATH, 'r', encoding='utf-8-sig') as f:
        sample = f.read(20000)

    try:
        sep = csv.Sniffer().sniff(
            sample, delimiters=';,\t|'
        ).delimiter
    except:
        sep = ';'

    df = pd.read_csv(
        FILE_PATH,
        sep=sep,
        encoding='utf-8-sig',
        engine='python',
        on_bad_lines='warn'
    )

df.columns = df.columns.astype(str).str.strip().str.lower()

required = [
    'date', 'message', 'views', 'comments',
    'reactions_total', 'forwards',
    'channel_members_count', 'topic_title'
]

for col in required:
    if col not in df.columns:
        df[col] = np.nan

df = df.copy()

df['message'] = df['message'].fillna('').astype(str)

for col in [
    'views', 'comments', 'reactions_total',
    'forwards', 'channel_members_count'
]:
    df[col] = pd.to_numeric(df[col], errors='coerce').fillna(0)

df['date'] = pd.to_datetime(
    df['date'], errors='coerce', utc=True
)

df = df.dropna(subset=['date']).copy()

df['topic_title'] = (
    df['topic_title'].fillna('Не указано').astype(str)
)

df['engagement'] = (
    df['reactions_total'] + df['comments']
)

df['ER_percent'] = np.where(
    df['views'] > 0,
    df['engagement'] / df['views'] * 100,
    np.nan
)

df['hour'] = df['date'].dt.hour
df['weekday'] = df['date'].dt.dayofweek
df['month_name'] = df['date'].dt.strftime('%Y-%m')
df['text_length'] = df['message'].str.len()
df['word_count'] = df['message'].apply(
    lambda x: len(re.findall(r'\b\w+\b', x))
)

# Удаляем строки, которые не являются текстовыми публикациями,
# только для анализа текста; исходный df не меняем
text_df = df[df['message'].str.strip().ne('')].copy()

print('Количество публикаций:', len(df))
print('Количество публикаций с текстом:', len(text_df))


# ============================================================
# 1. КАТЕГОРИЗАЦИЯ ТЕМАТИКИ ПУБЛИКАЦИЙ
# ============================================================

print('\n1. АНАЛИЗ ТЕМАТИКИ')

# Словари ключевых слов можно адаптировать под содержание канала
categories = {
    'Учебный процесс': [
        'учеб', 'занят', 'лекци', 'экзамен',
        'обучен', 'образован', 'курсант', 'студент'
    ],
    'Спорт': [
        'спорт', 'соревнован', 'турнир', 'футбол',
        'волейбол', 'баскетбол', 'побед', 'трениров'
    ],
    'Поступление': [
        'поступлен', 'абитуриент', 'прием',
        'приём', 'специальност', 'вступительн'
    ],
    'Профессиональная деятельность': [
        'полици', 'сотрудник', 'служб',
        'правоохран', 'оперативн', 'безопасност'
    ],
    'Памятные даты и поздравления': [
        'поздрав', 'праздник', 'юбилей',
        'день победы', 'памят', 'годовщин'
    ],
    'Научная деятельность': [
        'научн', 'конференц', 'исследован',
        'стать', 'доклад', 'форум'
    ],
    'Воспитательная и патриотическая работа': [
        'патриот', 'ветеран', 'истори',
        'геро', 'музей', 'память'
    ],
    'Мероприятия и события': [
        'мероприят', 'встреч', 'экскурс',
        'церемони', 'открыти', 'участ'
    ]
}

def classify_topic(text):
    text = text.lower()
    scores = {}

    for category, keywords in categories.items():
        scores[category] = sum(
            1 for word in keywords if word in text
        )

    best_category = max(scores, key=scores.get)

    if scores[best_category] == 0:
        return 'Прочее / не определено'

    return best_category

df['auto_category'] = df['message'].apply(classify_topic)

topic_comparison = (
    df.groupby('auto_category')
    .agg(
        publications=('views', 'count'),
        average_views=('views', 'mean'),
        median_views=('views', 'median'),
        average_reactions=('reactions_total', 'mean'),
        average_comments=('comments', 'mean'),
        average_forwards=('forwards', 'mean'),
        average_ER=('ER_percent', 'mean')
    )
    .round(2)
    .sort_values('publications', ascending=False)
    .reset_index()
)

display(topic_comparison)

plt.figure(figsize=(13, 6))
sns.barplot(
    data=topic_comparison,
    x='auto_category',
    y='average_views'
)
plt.title('Средние просмотры по тематическим категориям')
plt.xticks(rotation=60, ha='right')
plt.tight_layout()
plt.show()


# ============================================================
# 2. ВОПРОСЫ, ПРИЗЫВЫ К ДЕЙСТВИЮ И ЭМОДЗИ
# ============================================================

print('\n2. АНАЛИЗ ЭЛЕМЕНТОВ ТЕКСТА')

df['has_question'] = df['message'].str.contains(
    r'\?', regex=True
)

cta_pattern = (
    r'подписывайтесь|переходите|читайте|'
    r'смотрите|присоединяйтесь|участвуйте|'
    r'узнайте|поделитесь|оставляйте комментарии'
)

df['has_CTA'] = df['message'].str.contains(
    cta_pattern, case=False, regex=True
)

df['has_number'] = df['message'].str.contains(
    r'\d', regex=True
)

# Приближённое определение эмодзи
emoji_pattern = (
    r'[\U0001F300-\U0001FAFF'
    r'\U00002600-\U000027BF]'
)

df['has_emoji'] = df['message'].str.contains(
    emoji_pattern, regex=True
)

df['has_hashtag'] = df['message'].str.contains(
    r'#[а-яёa-z0-9_]+', case=False, regex=True
)

def compare_binary_feature(column, metric):
    result = (
        df.groupby(column)
        .agg(
            publications=(metric, 'count'),
            average=(metric, 'mean'),
            median=(metric, 'median')
        )
        .round(2)
        .reset_index()
    )

    groups = [
        group[metric].dropna().values
        for _, group in df.groupby(column)
        if len(group[metric].dropna()) >= 2
    ]

    if len(groups) == 2:
        u_stat, p_value = stats.mannwhitneyu(
            groups[0], groups[1],
            alternative='two-sided'
        )
    else:
        u_stat, p_value = np.nan, np.nan

    return result, p_value

feature_results = []

for feature in [
    'has_question', 'has_CTA',
    'has_number', 'has_emoji', 'has_hashtag'
]:
    result, p_value = compare_binary_feature(
        feature, 'views'
    )

    result.insert(0, 'Признак', feature)
    result['p_value'] = p_value

    feature_results.append(result)

feature_analysis = pd.concat(
    feature_results, ignore_index=True
)

display(feature_analysis)


# ============================================================
# 3. АНАЛИЗ РЕГУЛЯРНОСТИ ПУБЛИКАЦИЙ
# ============================================================

print('\n3. РЕГУЛЯРНОСТЬ ПУБЛИКАЦИЙ')

df = df.sort_values('date').copy()

df['previous_date'] = df['date'].shift(1)

df['interval_hours'] = (
    df['date'] - df['previous_date']
).dt.total_seconds() / 3600

interval_stats = pd.DataFrame({
    'Показатель': [
        'Средний интервал, часов',
        'Медианный интервал, часов',
        'Минимальный интервал, часов',
        'Максимальный интервал, часов',
        'Среднее количество публикаций в день'
    ],
    'Значение': [
        df['interval_hours'].mean(),
        df['interval_hours'].median(),
        df['interval_hours'].min(),
        df['interval_hours'].max(),
        df.groupby(df['date'].dt.date).size().mean()
    ]
})

display(interval_stats.round(2))

daily_stats = (
    df.groupby(df['date'].dt.date)
    .agg(
        publications=('views', 'count'),
        average_views=('views', 'mean'),
        total_views=('views', 'sum')
    )
    .reset_index()
)

daily_stats.columns = [
    'date', 'publications',
    'average_views', 'total_views'
]

display(daily_stats.head(20))

plt.figure(figsize=(14, 5))
plt.plot(
    daily_stats['date'],
    daily_stats['publications']
)
plt.title('Количество публикаций по дням')
plt.xlabel('Дата')
plt.ylabel('Количество публикаций')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()

# Проверка связи интервала и просмотров
interval_valid = df.dropna(
    subset=['interval_hours', 'views']
)

if len(interval_valid) >= 3:
    interval_corr, interval_p = stats.spearmanr(
        interval_valid['interval_hours'],
        interval_valid['views']
    )

    print('Корреляция интервала и просмотров:', round(interval_corr, 3))
    print('p-value:', interval_p)


# ============================================================
# 4. ВЫЯВЛЕНИЕ АНОМАЛЬНО ПОПУЛЯРНЫХ ПУБЛИКАЦИЙ
# ============================================================

print('\n4. АНОМАЛЬНО ПОПУЛЯРНЫЕ ПУБЛИКАЦИИ')

# Метод IQR
q1 = df['views'].quantile(0.25)
q3 = df['views'].quantile(0.75)
iqr = q3 - q1

upper_limit = q3 + 1.5 * iqr

df['is_view_outlier'] = df['views'] > upper_limit

outlier_posts = df[
    df['is_view_outlier']
][
    [
        'date', 'message', 'views',
        'reactions_total', 'comments',
        'forwards', 'auto_category'
    ]
].sort_values('views', ascending=False)

print('Порог аномально высоких просмотров:', upper_limit)
print('Количество аномально популярных публикаций:', len(outlier_posts))

display(outlier_posts.head(30))


# ============================================================
# 5. TF-IDF: ОСОБЕННОСТИ ПОПУЛЯРНЫХ ПУБЛИКАЦИЙ
# ============================================================

print('\n5. TF-IDF АНАЛИЗ')

# Исключаем пустые и слишком короткие тексты
tfidf_df = df[df['message'].str.len() >= 20].copy()

# Делим публикации по медиане просмотров
median_views = tfidf_df['views'].median()

tfidf_df['popularity_group'] = np.where(
    tfidf_df['views'] >= median_views,
    'Высокие просмотры',
    'Низкие просмотры'
)

def get_top_tfidf_words(data, top_n=25):
    if len(data) < 2:
        return pd.DataFrame(columns=['Слово', 'Средний TF-IDF'])

    vectorizer = TfidfVectorizer(
        lowercase=True,
        token_pattern=r'(?u)\b[а-яёa-z]{3,}\b',
        max_features=3000,
        stop_words=None
    )

    matrix = vectorizer.fit_transform(data['message'])

    average_scores = np.asarray(
        matrix.mean(axis=0)
    ).ravel()

    words = vectorizer.get_feature_names_out()

    result = pd.DataFrame({
        'Слово': words,
        'Средний TF-IDF': average_scores
    })

    return result.sort_values(
        'Средний TF-IDF',
        ascending=False
    ).head(top_n)

high_words = get_top_tfidf_words(
    tfidf_df[
        tfidf_df['popularity_group'] == 'Высокие просмотры'
    ]
)

low_words = get_top_tfidf_words(
    tfidf_df[
        tfidf_df['popularity_group'] == 'Низкие просмотры'
    ]
)

print('\nХарактерные слова публикаций с высокими просмотрами:')
display(high_words)

print('\nХарактерные слова публикаций с низкими просмотрами:')
display(low_words)


# ============================================================
# 6. КЛАСТЕРИЗАЦИЯ ПУБЛИКАЦИЙ
# ============================================================

print('\n6. КЛАСТЕРИЗАЦИЯ K-MEANS')

cluster_df = df[
    df['message'].str.len() >= 20
].copy()

if len(cluster_df) >= 10:

    vectorizer = TfidfVectorizer(
        max_features=500,
        min_df=2,
        token_pattern=r'(?u)\b[а-яёa-z]{3,}\b'
    )

    X_text = vectorizer.fit_transform(
        cluster_df['message']
    )

    # 4 группы публикаций
    n_clusters = min(4, len(cluster_df) - 1)

    model = KMeans(
        n_clusters=n_clusters,
        random_state=42,
        n_init=10
    )

    cluster_df['cluster'] = model.fit_predict(X_text)

    cluster_stats = (
        cluster_df.groupby('cluster')
        .agg(
            publications=('views', 'count'),
            average_views=('views', 'mean'),
            median_views=('views', 'median'),
            average_reactions=('reactions_total', 'mean'),
            average_comments=('comments', 'mean'),
            average_forwards=('forwards', 'mean'),
            average_text_length=('text_length', 'mean')
        )
        .round(2)
        .reset_index()
    )

    display(cluster_stats)

    terms = vectorizer.get_feature_names_out()

    print('\nКлючевые слова кластеров:')

    for cluster_number, center in enumerate(model.cluster_centers_):
        top_indices = center.argsort()[-10:][::-1]

        top_terms = [
            terms[i] for i in top_indices
        ]

        print(
            f'Кластер {cluster_number}:',
            ', '.join(top_terms)
        )

    print('\nПримеры публикаций по кластерам:')

    for cluster_number in sorted(cluster_df['cluster'].unique()):
        print(f'\nКЛАСТЕР {cluster_number}')

        display(
            cluster_df[
                cluster_df['cluster'] == cluster_number
            ][
                ['message', 'views', 'reactions_total']
            ].head(3)
        )

else:
    cluster_stats = pd.DataFrame()
    print('Недостаточно текстовых публикаций для кластеризации.')


# ============================================================
# 7. РЕГРЕССИОННЫЙ АНАЛИЗ
# ============================================================

print('\n7. РЕГРЕССИОННЫЙ АНАЛИЗ')

# Проверяем связь просмотров с характеристиками публикации.
# Используем только признаки, доступные в таблице.

regression_df = df.copy()

regression_df['log_views'] = np.log1p(
    regression_df['views']
)

regression_df['log_reactions'] = np.log1p(
    regression_df['reactions_total']
)

regression_df['log_comments'] = np.log1p(
    regression_df['comments']
)

regression_df['log_forwards'] = np.log1p(
    regression_df['forwards']
)

regression_df['has_question_num'] = (
    regression_df['message'].str.contains(r'\?')
).astype(int)

regression_df['has_emoji_num'] = (
    regression_df['message'].str.contains(
        emoji_pattern, regex=True
    )
).astype(int)

regression_df['has_CTA_num'] = (
    regression_df['message'].str.contains(
        cta_pattern, case=False, regex=True
    )
).astype(int)

features = [
    'text_length',
    'word_count',
    'hour',
    'weekday',
    'has_question_num',
    'has_emoji_num',
    'has_CTA_num'
]

X = regression_df[features]
y = regression_df['log_views']

# Удаляем строки без целевой переменной
valid = y.notna() & np.isfinite(y)

X = X.loc[valid]
y = y.loc[valid]

if len(X) >= 20:

    # Простая модель Random Forest
    # Важно: это анализ предсказательной важности,
    # а не доказательство причинного влияния.

    pipeline = Pipeline([
        ('imputer', SimpleImputer(strategy='median')),
        ('model', RandomForestRegressor(
            n_estimators=300,
            random_state=42,
            min_samples_leaf=3
        ))
    ])

    pipeline.fit(X, y)

    predictions = pipeline.predict(X)

    print('R² на обучающей выборке:',
          round(r2_score(y, predictions), 3))

    print('MAE на обучающей выборке:',
          round(mean_absolute_error(y, predictions), 3))

    importance = pd.DataFrame({
        'Признак': features,
        'Важность': pipeline.named_steps[
            'model'
        ].feature_importances_
    }).sort_values(
        'Важность', ascending=False
    )

    display(importance)

    plt.figure(figsize=(10, 5))
    sns.barplot(
        data=importance,
        x='Важность',
        y='Признак'
    )
    plt.title('Важность характеристик для модели просмотров')
    plt.tight_layout()
    plt.show()

else:
    importance = pd.DataFrame()
    print('Недостаточно данных для регрессионного анализа.')


# ============================================================
# 8. АНАЛИЗ ТОНАЛЬНОСТИ
# ============================================================

print('\n8. ПРИБЛИЖЁННЫЙ АНАЛИЗ ТОНАЛЬНОСТИ')

positive_words = [
    'победа', 'успех', 'поздравляем', 'достижение',
    'благодарность', 'гордимся', 'лучший',
    'победитель', 'праздник', 'успешно'
]

negative_words = [
    'нарушение', 'преступление', 'опасность',
    'трагедия', 'происшествие', 'угроза',
    'задержание', 'погиб', 'авария', 'вред'
]

def sentiment_score(text):
    text = text.lower()

    positive = sum(
        text.count(word) for word in positive_words
    )

    negative = sum(
        text.count(word) for word in negative_words
    )

    return positive - negative

df['sentiment_score'] = df['message'].apply(
    sentiment_score
)

df['sentiment_category'] = np.select(
    [
        df['sentiment_score'] > 0,
        df['sentiment_score'] < 0
    ],
    [
        'Условно положительная',
        'Условно отрицательная'
    ],
    default='Нейтральная'
)

sentiment_stats = (
    df.groupby('sentiment_category')
    .agg(
        publications=('views', 'count'),
        average_views=('views', 'mean'),
        average_reactions=('reactions_total', 'mean'),
        average_comments=('comments', 'mean'),
        average_ER=('ER_percent', 'mean')
    )
    .round(2)
    .reset_index()
)

display(sentiment_stats)

print(
    'Внимание: это простой словарный индикатор, '
    'а не полноценная модель анализа тональности.'
)


# ============================================================
# 9. АНАЛИЗ ЗАГОЛОВКОВ
# ============================================================

print('\n9. АНАЛИЗ ЗАГОЛОВКОВ')

df['first_sentence'] = (
    df['message']
    .str.split(r'[.!?\n]')
    .str[0]
)

df['first_sentence_length'] = (
    df['first_sentence'].str.len()
)

df['headline_question'] = (
    df['first_sentence'].str.contains(r'\?', na=False)
)

df['headline_exclamation'] = (
    df['first_sentence'].str.contains(r'!', na=False)
)

df['headline_number'] = (
    df['first_sentence'].str.contains(r'\d', na=False)
)

headline_stats = (
    df.groupby(
        [
            'headline_question',
            'headline_exclamation',
            'headline_number'
        ]
    )
    .agg(
        publications=('views', 'count'),
        average_views=('views', 'mean'),
        average_ER=('ER_percent', 'mean')
    )
    .round(2)
    .reset_index()
)

display(headline_stats)

# Связь длины первого предложения с просмотрами
headline_corr, headline_p = stats.spearmanr(
    df['first_sentence_length'],
    df['views'],
    nan_policy='omit'
)

print('Корреляция длины первого предложения и просмотров:',
      round(headline_corr, 3))
print('p-value:', headline_p)


# ============================================================
# 10. СРАВНЕНИЕ ПОКАЗАТЕЛЕЙ ПО МЕСЯЦАМ
# ============================================================

print('\n10. ИЗМЕНЕНИЕ ЭФФЕКТИВНОСТИ КАНАЛА')

monthly_advanced = (
    df.groupby('month_name')
    .agg(
        publications=('views', 'count'),
        average_views=('views', 'mean'),
        median_views=('views', 'median'),
        average_reactions=('reactions_total', 'mean'),
        average_comments=('comments', 'mean'),
        average_forwards=('forwards', 'mean'),
        average_ER=('ER_percent', 'mean'),
        average_text_length=('text_length', 'mean')
    )
    .round(2)
    .reset_index()
)

display(monthly_advanced)

plt.figure(figsize=(14, 5))
plt.plot(
    monthly_advanced['month_name'],
    monthly_advanced['average_ER'],
    marker='o'
)
plt.title('Динамика ER по месяцам')
plt.xlabel('Месяц')
plt.ylabel('ER, %')
plt.xticks(rotation=45)
plt.grid(alpha=0.3)
plt.tight_layout()
plt.show()


# ============================================================
# 11. ЭКСПОРТ ВСЕХ ДОПОЛНИТЕЛЬНЫХ РЕЗУЛЬТАТОВ
# ============================================================

print('\n11. СОХРАНЕНИЕ РЕЗУЛЬТАТОВ')

OUTPUT_FILE = '/content/mvd_advanced_analysis.xlsx'

sheets = {
    'Категории': topic_comparison,
    'Элементы текста': feature_analysis,
    'Интервалы': interval_stats,
    'Дневная активность': daily_stats,
    'Аномальные посты': outlier_posts,
    'TFIDF высокие': high_words,
    'TFIDF низкие': low_words,
    'Кластеры': cluster_stats,
    'Важность признаков': importance,
    'Тональность': sentiment_stats,
    'Заголовки': headline_stats,
    'Динамика': monthly_advanced
}

df_export = df.copy()

for col in df_export.columns:
    if isinstance(df_export[col].dtype, pd.DatetimeTZDtype):
        df_export[col] = df_export[col].dt.tz_localize(None)

with pd.ExcelWriter(
    OUTPUT_FILE,
    engine='openpyxl'
) as writer:

    for sheet_name, data in sheets.items():
        if isinstance(data, pd.DataFrame):
            data.to_excel(
                writer,
                sheet_name=sheet_name[:31],
                index=False
            )

    df_export.to_excel(
        writer,
        sheet_name='Публикации с метриками',
        index=False
    )

print('\nАНАЛИЗ ЗАВЕРШЁН!')
print('Файл:', OUTPUT_FILE)

from google.colab import files
files.download(OUTPUT_FILE)
```
